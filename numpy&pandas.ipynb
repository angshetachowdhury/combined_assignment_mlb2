## Question 4: Indexing, slicing, stacking and splitting

**Tasks**

- Create one 1D, one 2D and one 3D array, and write at least 5 indexing or slicing examples for each.
- Using the arrays d and e given below, join them with hstack and vstack.
- Split the joined results back using hsplit and vsplit.

**How to do it**

For indexing, remember counting starts at zero and negative numbers count from the end. For slicing, the pattern is start, stop and step, and the stop position is not included. In 2D arrays, you give one slice for rows and one for columns. In 3D arrays, you give one choice for each of the three axes. Include a mix of examples: a single element, a whole row or column, a range, a step, and a reverse order. You can also try selecting by a condition. For stacking, vstack puts one array below the other (rows add up) and hstack puts them side by side (columns add up); predict the new shape before you run it. For splitting, the array must divide evenly into the number of parts you ask for, so choose the number of parts according to the number of columns or rows.

```python
import numpy as np

# ==========================================
# 1. Create 1D, 2D and 3D arrays
# ==========================================

array_1d = np.arange(10)

array_2d = np.arange(20).reshape(4, 5)

array_3d = np.arange(24).reshape(2, 3, 4)

print("1D Array:")
print(array_1d)

print("\n2D Array:")
print(array_2d)

print("\n3D Array:")
print(array_3d)


# ==========================================
# 2. Indexing and slicing - 1D array
# ==========================================

print("\n========== 1D INDEXING / SLICING ==========")

print("1. First element:", array_1d[0])

print("2. Last element:", array_1d[-1])

print("3. Elements from index 2 to 5:", array_1d[2:6])

print("4. Every second element:", array_1d[::2])

print("5. Reverse order:", array_1d[::-1])

print("6. Elements greater than 5:", array_1d[array_1d > 5])


# ==========================================
# 3. Indexing and slicing - 2D array
# ==========================================

print("\n========== 2D INDEXING / SLICING ==========")

print("1. First element:", array_2d[0, 0])

print("2. Last element:", array_2d[-1, -1])

print("3. First row:", array_2d[0, :])

print("4. First column:", array_2d[:, 0])

print("5. Rows 1 to 2 and columns 1 to 3:")
print(array_2d[1:3, 1:4])

print("6. Every second column:")
print(array_2d[:, ::2])


# ==========================================
# 4. Indexing and slicing - 3D array
# ==========================================

print("\n========== 3D INDEXING / SLICING ==========")

print("1. First element:", array_3d[0, 0, 0])

print("2. Last element:", array_3d[-1, -1, -1])

print("3. First 2D block:")
print(array_3d[0])

print("4. First row from every block:")
print(array_3d[:, 0, :])

print("5. First column from every row:")
print(array_3d[:, :, 0])

print("6. Every second element along the last axis:")
print(array_3d[:, :, ::2])

print("7. Reverse the last axis:")
print(array_3d[:, :, ::-1])


# ==========================================
# 5. Given arrays d and e
# ==========================================

d = np.arange(27).reshape(3, 9)

e = np.arange(28, 55).reshape(3, 9)

print("\n========== d AND e ==========")

print("d:")
print(d)

print("\ne:")
print(e)


# ==========================================
# 6. Horizontal stacking
# ==========================================

hstack_result = np.hstack((d, e))

print("\n========== HSTACK ==========")
print(hstack_result)
print("Shape:", hstack_result.shape)


# ==========================================
# 7. Vertical stacking
# ==========================================

vstack_result = np.vstack((d, e))

print("\n========== VSTACK ==========")
print(vstack_result)
print("Shape:", vstack_result.shape)


# ==========================================
# 8. Split hstack result using hsplit
# ==========================================

hstack_split = np.hsplit(hstack_result, 2)

print("\n========== HSPLIT ==========")

print("First part:")
print(hstack_split[0])

print("\nSecond part:")
print(hstack_split[1])


# ==========================================
# 9. Split vstack result using vsplit
# ==========================================

vstack_split = np.vsplit(vstack_result, 2)

print("\n========== VSPLIT ==========")

print("First part:")
print(vstack_split[0])

print("\nSecond part:")
print(vstack_split[1])
```



observation: The 1D, 2D and 3D arrays demonstrate indexing and slicing using positive and negative indexes, ranges, steps and reverse order. Boolean indexing was also used to select values based on a condition.

The two arrays d and e have shape (3, 9). Horizontal stacking produces a (3, 18) array, while vertical stacking produces a (6, 9) array. The stacked arrays can be divided back into two equal parts using hsplit and vsplit.
